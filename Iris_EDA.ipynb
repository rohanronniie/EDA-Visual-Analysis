# Iris Dataset – Exploratory Data Analysis
This notebook performs exploratory data analysis on the Iris dataset to study feature distributions, species balance, and correlations.

## Importing Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading the Dataset

```python
iris = pd.read_csv("Iris.csv")
iris.head()
```

## Dataset Structure and Data Types

```python
iris.info()
```

## Feature Distributions

```python
iris.hist(figsize=(6,8))
plt.show()
```

## Species Distribution

```python
sns.countplot(x='Species', data=iris)
plt.show()
```

## Outlier Detection

```python
sns.boxplot(data=iris.iloc[:, 1:-1])
plt.xticks(rotation=90)
plt.ylabel("value")
plt.show()
```

## Correlation Heatmap

```python
plt.figure(figsize=(6,4))
sns.heatmap(iris.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

## Final Summary

- The dataset is evenly distributed across all three Iris species.
- Petal features show greater variation compared to sepal features.
- Petal Length and Petal Width are highly correlated and are the most important features for classification.
- Sepal Width shows weaker relationships with other features.
- Only mild outliers are observed, mainly in Sepal Width.
- Iris-setosa is clearly distinguishable from the other two species.
